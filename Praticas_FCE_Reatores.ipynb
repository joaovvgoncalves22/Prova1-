```python
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt

@dataclass


class ParametrosCSTR:
    F: float # Vazão (L/min)
    V: float # Volume (L)
    k: float # Constante cinética (min^-1)

# Modelo de um único CSTR
def modelo_cstr(CAin, params):

    tau = params.V / params.F

    CA = CAin / (1 + params.k * tau)

    return CA

  # Simulação dos CSTRs em série
def simula_cstr(CA0, params, num_estagios=3):

    concentracoes = []
    conversoes = []

    CAentrada = CA0

    for i in range(num_estagios):


        CAsaida = modelo_cstr(CAentrada, params)


        X = (CA0 - CAsaida) / CA0

        concentracoes.append(CAsaida)
        conversoes.append(X)


        CAentrada = CAsaida

    return concentracoes, conversoes
```
