###(c) Descreva as etapas de entrada, processamento e saída do algoritmo.

```python
def controle_tanque():
    limite_min = 20
    limite_max = 80
    cont_baixo = 0

    while True:
        entrada = input("Digite o nível (0 a 100) ou sair: ")

        # opção pra encerrar
        if entrada.lower() == 'sair':
            print("Programa encerrado")
            break

        try:
            nivel = float(entrada)

            # checando se está dentro do intervalo
            if nivel < 0 or nivel > 100:
                print("Valor inválido (0 a 100)")
                continue

            if nivel < limite_min:
                print("Nível baixo")
                cont_baixo += 1

            elif nivel <= limite_max:
                print("Operação normal")
                cont_baixo = 0

            else:
                print("Risco de transbordamento")
                cont_baixo = 0

            # se deu 3 vezes seguidas
            if cont_baixo == 3:
                print("ALERTA: parar sistema")
                break

        except:
            print("Erro na entrada")

controle_tanque()
```

#Questão 4
###Durante a operação de um processo, a concentração de um reagente é monitorada periodicamente para avaliar as condições de operação do sistema. A concentração pode variar ao longo do tempo e deve ser classificada em diferentes faixas para auxiliar na análise do processo. Neste contexto, você recebeu a tarefa de criar um algoritmo que permita executar a simulação do sistema de monitoramento, o qual deve seguir os seguintes requisitos:

• A cada leitura, deve ser gerado um valor aleatório de concentração entre 0 e 10 mol/L.
• Se a concentração for menor que 3, o sistema deve indicar “Baixa”.
• Se a concentração estiver entre 3 e 7, o sistema deve indicar “Adequada”.
• Se a concentração for maior que 7, o sistema deve indicar “Alta”.
• Ao final da simulação, deve-se contabilizar quantas vezes cada condição ocorreu.
Com base nesses requisitos, desenvolva um código modularizado em Python para simular o
monitoramento considerando 10 leituras consecutivas.
O código deve conter, obrigatoriamente:
• Pelo menos uma função
• Estrutura de repetição
• Estrutura condicional
• Comentários explicativos

Ao final da execução, o programa deve apresentar:
• O número de ocorrências de cada classe (Baixa, Adequada e Alta)
• Uma mensagem indicando qual condição foi predominante no processo

```python
import numpy as np

# função para classificar o valor da concentração
def classificar(valor):
    if valor < 3:
        return "Baixa"
    elif valor <= 7:
        return "Adequada"
    else:
        return "Alta"


# função que faz a simulação das leituras
def simular(n):
    cont = {"Baixa": 0, "Adequada": 0, "Alta": 0}

    for i in range(n):
        c = np.random.uniform(0, 10)  # gera valor aleatório entre 0 e 10
        situacao = classificar(c)     # classifica o valor gerado

        print(f"Leitura {i+1}: Concentração = {c:.1f} mol/L -> {situacao}")

        cont[situacao] += 1  # soma no contador correspondente

    return cont


# função para mostrar o resumo final
def mostrar_resumo(cont):
    print("\nResumo:")
    print(f"Baixa: {cont['Baixa']} ocorrências")
    print(f"Adequada: {cont['Adequada']} ocorrências")
    print(f"Alta: {cont['Alta']} ocorrências")

    # mostra qual apareceu mais vezes
    print(f"\nCondição predominante: {max(cont, key=cont.get)}")


# executa o programa
dados = simular(10)
mostrar_resumo(dados)
```
