```python
from dataclasses import dataclass

@dataclass
class Corrente:

  vazão: float
  fração: float
```

```python
def misturador(c1: Corrente, c2: Corrente):
  """
  Calcule a vazão e a fração da mistura de  correntes

  """


  #Primeira Corrente
  Fa = c1.vazão
  xa =  c1.fração

  #Segunda Corrente
  Fb = c2.vazão
  xb =  c2.fração

  #Vazão Total
  F = Fa + Fb

  x =(Fa*xa + Fb*xb)/(F)


  return F, x
```
